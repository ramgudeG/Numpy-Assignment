**Q1.** Create the array `temperatures = [36.5, 37.2, 38.9, 36.1, 39.5, 37.8, 40.1, 36.8]`.  
- Find all temperatures **above 38.0** (fever threshold).  
- Count how many readings are **normal** (between 36.0 and 37.5).  
- What percentage of readings are **high fever** (above 39.0)?

**Q2.** Given a dataset of house prices (in thousands):  
`prices = [120, 450, 85, 320, 680, 95, 210, 540, 75, 390]`  
- Select prices that are **affordable** (below 200k).  
- Select prices that are **luxury** (above 400k).  
- Replace all prices above 500k with exactly `500` (cap at 500).

**Q3.** You have exam scores for 5 students across 3 subjects:
```python
exams = np.array([[55, 72, 80],
                  [90, 85, 92],
                  [40, 55, 60],
                  [78, 88, 75],
                  [30, 45, 50]])
```
- Find students (rows) where **all three scores are above 70**.  
- Use `np.where` to assign grade labels: `'A'` if average ≥ 80, else `'B'`.

````python
# 1.

temperatures = np.array([36.5, 37.2, 38.9, 36.1, 39.5, 37.8, 40.1, 36.8])

# Temperatures above 38.0
fever = temperatures[temperatures > 38.0]
print("Above 38.0:", fever)

# Normal temperatures (36.0 to 37.5 inclusive)
normal_mask = (temperatures >= 36.0) & (temperatures <= 37.5)
normal_count = np.sum(normal_mask)
print("Normal count:", normal_count)

# High fever (> 39.0)
high_fever_count = np.sum(temperatures > 39.0)
percentage = (high_fever_count / len(temperatures)) * 100
print("High fever percentage:", percentage)
````

````python
# 2.
prices = np.array([120, 450, 85, 320, 680, 95, 210, 540, 75, 390])

# Affordable houses (<200k)
affordable = prices[prices < 200]
print("Affordable:", affordable)

# Luxury houses (>400k)
luxury = prices[prices > 400]
print("Luxury:", luxury)

# Cap prices at 500k
capped_prices = np.where(prices > 500, 500, prices)
print("Capped prices:", capped_prices)
````

````python
# 3.
exams = np.array([
    [55, 72, 80],
    [90, 85, 92],
    [40, 55, 60],
    [78, 88, 75],
    [30, 45, 50]
])

# Students with all scores > 70
mask = np.all(exams > 70, axis=1)

print("Students with all scores > 70:")
print(exams[mask])

# Average score per student
averages = exams.mean(axis=1)

# Grade assignment
grades = np.where(averages >= 80, 'A', 'B')

print("Averages:", averages)
print("Grades:", grades)
````

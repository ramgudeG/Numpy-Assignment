**Q28.**

- Use `np.argsort` to get indices that would sort the array ascending
- Use those indices (fancy indexing) to produce the sorted array
- Use `np.argsort` on the result of `np.argsort` to assign ranks (1-based, 1 = smallest)

````python
import numpy as np

scores = np.array([72, 55, 91, 38, 85, 63, 78, 44])

# Indices that would sort the array
sort_indices = np.argsort(scores)

# Sorted array using fancy indexing
sorted_scores = scores[sort_indices]

# Ranks (1 = smallest)
ranks = np.argsort(np.argsort(scores)) + 1

print("Original Scores:", scores)
print("Sort Indices:", sort_indices)
print("Sorted Scores:", sorted_scores)
print("Ranks:", ranks)
````

````python
import numpy as np
import time

arr64 = np.random.rand(1000,1000)

arr32 = arr64.astype(np.float32)

print("float64 bytes:", arr64.nbytes)
print("float32 bytes:", arr32.nbytes)

start = time.perf_counter()
s64 = arr64.sum()
t64 = time.perf_counter() - start

start = time.perf_counter()
s32 = arr32.sum()
t32 = time.perf_counter() - start

print("\nMeans:")
print(arr64.mean())
print(arr32.mean())

print("\nTimes:")
print("float64:", t64)
print("float32:", t32)
````
